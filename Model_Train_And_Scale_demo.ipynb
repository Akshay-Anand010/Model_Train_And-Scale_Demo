```python
import pandas as pd
```

```python
red_wine=pd.read_csv('/content/sample_data/winequality-red.csv')
```

```python
red_wine.head()
```

```python
red_wine.shape
```

```python
red_wine1=pd.read_csv('/content/sample_data/winequality-red.csv',sep=';')
```

```python
red_wine1.head()
```

```python
red_wine1.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(red_wine1.drop('quality',axis=1),red_wine1['quality'])
```

```python
from sklearn.svm import SVC
```

```python
clf = SVC(kernel='linear')
```

```python
clf.fit(x_train,y_train)
```

```python
y_pred=clf.predict(x_test)
```

```python
print(y_pred)
```

```python
from sklearn import metrics
```

```python
metrics.accuracy_score(y_test,y_pred)
```

```python
import matplotlib.pyplot as plot
```

```python
plot.plot(y_test,y_pred)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaled_data=scaler.fit_transform(red_wine1.drop('quality',axis=1))
```

```python
x_train,x_test,y_train,y_test=train_test_split(scaled_data,red_wine1['quality'], train_size=0.80)
```

```python
clf.fit(x_train,y_train)
```

```python
y_pred=clf.predict(x_test)
```

```python
print(y_pred)
```

```python
metrics.accuracy_score(y_test,y_pred)
```
